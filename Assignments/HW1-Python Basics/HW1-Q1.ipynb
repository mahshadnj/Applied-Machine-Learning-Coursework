Question 1:
Find the 2 largest numbers between the input numbers.

```python
print("Enter numbers between 10 and 90 (enter -1 to finish):")

nums = []
n = 0

while n != -1:
    n = int(input("Number: "))
    
    if n == -1:
        break
    elif n < 10 or n > 90:
        print("Please enter a number between 10 and 90.")
    else:
        nums.append(n)
        print("Current valid numbers:", nums)

# Find the largest and second largest numbers

max1 = nums[0]
max2 = nums[0]

for x in nums:
    if x > max1:
        max2 = max1
        max1 = x
    elif x > max2 and x != max1:
        max2 = x

print("\nLargest numbers are:")
print(max1, max2)
```
