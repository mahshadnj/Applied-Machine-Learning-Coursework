Question 3: Simple word translator using dictionary

```python
from collections import OrderedDict

print("Enter number of dictionary words:")

n = int(input("n: "))

words = OrderedDict()

print("\nEnter dictionary words:")
for i in range(n):
    print("two", i + 1, ":")
    line = input()
    two = line.split()
    words[two[0]] = two[1]
    print("Current dictionary:", words)

print("\nEnter a sentence to translate:")
sentence = input("Sentence: ").split()

translated = []

for word in sentence:
    if word in words:
        translated.append(words[word])
    else:
        translated.append(word)

print("\nTranslated sentence:")

output = ""
for w in translated:
    output += w + " "
print(output.strip())
```
