Question 4: Detect mirror strings

```python
n = int(input("Enter number of strings (n): "))

words = []

for i in range(n):
    s = input("Enter string " + str(i + 1) + ": ")
    words.append(s)

count = 0
nums = []

# Check each string
for i in range(n):
    word = words[i]
    reversed_word = ""
    for ch in word:
        reversed_word = ch + reversed_word
    if word == reversed_word:
        count += 1
        nums.append(i + 1)

# Show all input strings
print("\nList of entered strings:")
for i in range(n):
    print(i + 1, ":", words[i])

# Show results
print("\nTotal mirror strings:", count)
print("Their numbers are:")
for num in nums:
    print(num)
```
