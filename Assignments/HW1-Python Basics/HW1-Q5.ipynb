Question 5: Calculate averages from CSV data

```python
# Read CSV file
with open("grades.csv", "r") as file:
    lines = file.readlines()

names = []
averages = []

for line in lines:
    parts = line.strip().split(",")
    name = parts[0]
    scores = parts[1:]
    
    total = 0
    for s in scores:
        total += float(s)
    avg = total / len(scores)
    
    names.append(name)
    averages.append(avg)

# Part 1- average in original order
print("\nAverage of each person (original order):")
for i in range(len(names)):
    print(names[i] + "," + str(averages[i]) + "\n")

# Part 2- average sorted ascending
nam_c = names[:]
avg_c = averages[:]

# Sort manually
for i in range(len(avg_c)):
    for j in range(i + 1, len(avg_c)):
        if avg_c[i] > avg_c[j]:
            # swap averages
            temp_avg = avg_c[i]
            avg_c[i] = avg_c[j]
            avg_c[j] = temp_avg
            # swap names
            temp_name = nam_c[i]
            nam_c[i] = nam_c[j]
            nam_c[j] = temp_name

print("\nAverages in ascending order:")
for i in range(len(nam_c)):
    print(nam_c[i] + "," + str(avg_c[i]))

# Part 3- Calculate overall average
overall_avg = sum(averages) / len(averages)

print("\nOverall average of all people:", str(overall_avg))
```
